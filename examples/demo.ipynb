# crs-sniffer demo

Guess the CRS of the sample files in this folder. In Colab, the setup cell installs the package and downloads the sample files from the repository.

```python
import os
import urllib.request

try:
    from crs_sniffer import sniff, read_file
except ImportError:  # e.g. a fresh Colab runtime
    import subprocess, sys
    subprocess.check_call([sys.executable, '-m', 'pip', 'install', '-q', 'git+https://github.com/AmmarYasser455/crs-sniffer'])
    from crs_sniffer import sniff, read_file

FILES = ['london_bng.csv', 'paris_lambert93.csv', 'berlin_wgs84.geojson']
BASE = 'examples' if os.path.isdir('examples') else 'crs_sniffer_examples'
os.makedirs(BASE, exist_ok=True)
for name in FILES:
    path = os.path.join(BASE, name)
    if not os.path.exists(path):
        url = 'https://raw.githubusercontent.com/AmmarYasser455/crs-sniffer/main/examples/' + name
        urllib.request.urlretrieve(url, path)
```

```python
london = read_file(os.path.join(BASE, 'london_bng.csv'))
matches = sniff(london)
print(london)
print(matches)
assert matches[0].epsg == 27700
```

```python
paris = read_file(os.path.join(BASE, 'paris_lambert93.csv'))
print(sniff(paris)[0])
assert sniff(paris)[0].epsg == 2154
```

```python
berlin = read_file(os.path.join(BASE, 'berlin_wgs84.geojson'))
print(sniff(berlin)[0])
assert sniff(berlin)[0].epsg == 4326
```
